```python
import pandas as pd
df = pd.read_csv("../data/nairobi_cost_of_living.csv")
df.head()
```

```python
df.head()

df.info()

df.describe()

df.shape

df.columns
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.sort_values("Rent", ascending=False)
```

```python
df["Food"].mean()
```

```python
df["Total"] = (
    df["Food"] +
    df["Rent"] +
    df["Transport"]
)
```

```python
df.sort_values("Total", ascending=False)
```

```python
df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x="Area",
    y="Rent"
)

plt.xticks(rotation=90)
plt.show()
```

```python
totals = [
    df["Food"].sum(),
    df["Rent"].sum(),
    df["Transport"].sum()
]

labels = ["Food","Rent","Transport"]

plt.pie(
    totals,
    labels=labels,
    autopct="%1.1f%%"
)

plt.show()
```

```python
sns.histplot(df["Food"])
```

```python
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True
)
```

```python
df.to_csv(
    "cleaned_data.csv",
    index=False
)
```

```python
import streamlit as st
import pandas as pd

df = pd.read_csv("cleaned_data.csv")

st.title("Kenya Cost of Living Dashboard")

st.write(df)
```

```python
area = st.sidebar.selectbox(
    "Area",
    df["Area"].unique()
)

filtered = df[df["Area"] == area]

st.write(filtered)
```

```python
st.bar_chart(filtered["Food"])
```

```python
st.metric(
    "Average Rent",
    filtered["Rent"].mean()
)
```

```python
st.metric(
    "Average Food",
    filtered["Food"].mean()
)
```

```python
st.metric(
    "Average Transport",
    filtered["Transport"].mean()
)
```

```python
st.set_page_config(
    page_title="Nairobi Cost Dashboard",
    layout="wide"
)
```
